# Wavelet decomposition and reconstruction

A single Haar transform separates approximation and detail coefficients. Use the same wavelet and boundary mode for reconstruction. Odd input lengths can require trimming the reconstructed signal.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.5.0"
#r "nuget: NDSharp.Frame, 2.5.0"
#r "nuget: NDSharp.Plot, 2.5.0"
#r "nuget: NDSharp.Image, 2.5.0"
#r "nuget: NDSharp.Wavelets, 2.5.0"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var signal = new double[] { 1, 2, 3, 4, 5, 6, 7, 8 };
var wavelet = Wavelet.Get("haar");
var (approximation, detail) = Dwt.Dwt1D(signal, wavelet);
Console.WriteLine(string.Join(", ", approximation));
Console.WriteLine(string.Join(", ", detail));
var restored = Dwt.Idwt1D(approximation, detail, wavelet);
if (signal.Zip(restored).Any(p => Math.Abs(p.First - p.Second) > 1e-10)) throw new Exception("Wavelet roundtrip mismatch");
```
