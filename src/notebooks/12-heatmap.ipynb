# Heatmaps and colorbars

ImageArtist maps numeric values through a named colormap. AddColorbar attaches a scale to a mappable artist.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.5.0"
#r "nuget: NDSharp.Frame, 2.5.0"
#r "nuget: NDSharp.Plot, 2.5.0"
#r "nuget: NDSharp.Image, 2.5.0"
#r "nuget: NDSharp.Wavelets, 2.5.0"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var figure = new Figure();
var axes = figure.AddAxes(GridCell.Single(1, 1, 0));
var image = new ImageArtist { Rows = 2, Cols = 3, Values = new double[] { 0, 1, 2, 3, 4, 5 }, Cmap = Colormap.Get("viridis") };
axes.Add(image);
figure.AddColorbar(image, label: "value");
File.WriteAllBytes("ndsharp-heatmap.png", figure.RenderPng());
Console.WriteLine(Path.GetFullPath("ndsharp-heatmap.png"));
```
