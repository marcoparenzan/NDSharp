# Slicing and shared storage

A basic slice shares its buffer with the original. Copy before editing if you need isolation. Strides are measured in elements, not bytes.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.5.0"
#r "nuget: NDSharp.Frame, 2.5.0"
#r "nuget: NDSharp.Plot, 2.5.0"
#r "nuget: NDSharp.Image, 2.5.0"
#r "nuget: NDSharp.Wavelets, 2.5.0"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var a = np.Reshape(np.Arange(0L, 12L), 3, 4);
var view = a.Get(new Slice(0, 2), new Slice(1, 3));
Console.WriteLine(view);
Console.WriteLine($"Shared buffer: {ReferenceEquals(a.Buffer, view.Buffer)}");
Console.WriteLine(np.Transpose(a));
var independent = view.Copy();
if (ReferenceEquals(independent.Buffer, a.Buffer)) throw new Exception("Copy must own storage");
```
