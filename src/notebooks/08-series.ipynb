# Label alignment and missing values

Series arithmetic aligns labels rather than row positions. Unmatched labels introduce missing values. Frame reductions skip missing values by default.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.8.1-net11-preview.1"
#r "nuget: NDSharp.Frame, 2.5.1-net11-preview.1"
#r "nuget: NDSharp.Plot, 2.6.3-net11-preview.2"
#r "nuget: NDSharp.Image, 2.5.1-net11-preview.1"
#r "nuget: NDSharp.Wavelets, 2.5.1-net11-preview.1"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var a = new Series(Column.FromLongs(new long[] { 1, 2 }), FIndex.OfStrings(new[] { "x", "y" }));
var b = new Series(Column.FromLongs(new long[] { 10, 20 }), FIndex.OfStrings(new[] { "y", "z" }));
Console.WriteLine(Formatter.SeriesRepr(Ops.Binary(BinOp.Add, a, b), new DisplayOptions()));
var values = Column.FromDoubles(new[] { 1.0, double.NaN, 3.0 });
Console.WriteLine(NDSharp.Frame.Reduce.Scalar("mean", values));
if (!Equals(NDSharp.Frame.Reduce.Scalar("count", values), 2L)) throw new Exception("Missing count mismatch");
```
