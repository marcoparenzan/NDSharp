# FFT and signal differences

Transform a small signal and invert it. FFT output is complex-valued. Diff computes adjacent differences, not a derivative scaled by sample spacing.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.
