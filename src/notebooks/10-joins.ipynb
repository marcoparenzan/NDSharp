# Outer joins

Join two tables by key. An outer join retains both sets of keys; missing integer payloads are promoted to floating point.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.5.0"
#r "nuget: NDSharp.Frame, 2.5.0"
#r "nuget: NDSharp.Plot, 2.5.0"
#r "nuget: NDSharp.Image, 2.5.0"
#r "nuget: NDSharp.Wavelets, 2.5.0"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var left = new DataFrame(new[] { Column.FromLongs(new long[] { 1, 2 }), Column.FromLongs(new long[] { 10, 20 }) }, FIndex.OfStrings(new[] { "key", "a" }));
var right = new DataFrame(new[] { Column.FromLongs(new long[] { 2, 3 }), Column.FromLongs(new long[] { 5, 6 }) }, FIndex.OfStrings(new[] { "key", "b" }));
var joined = Merge.Join(left, right, new Merge.Spec { LeftOn = new object?[] { "key" }, RightOn = new object?[] { "key" }, How = "outer" });
Console.WriteLine(Formatter.FrameRepr(joined));
if (joined.NRows != 3) throw new Exception("Join mismatch");
```
