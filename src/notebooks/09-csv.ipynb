# CSV to a DataFrame

Csv.Read accepts CSV text, not a path. Pass the inferred names as the DataFrame column index. Use File.ReadAllText for a real file.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.5.0"
#r "nuget: NDSharp.Frame, 2.5.0"
#r "nuget: NDSharp.Plot, 2.5.0"
#r "nuget: NDSharp.Image, 2.5.0"
#r "nuget: NDSharp.Wavelets, 2.5.0"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var csv = "city,value\nParis,10\nRome,20\n";
var (columns, names) = Csv.Read(csv, new Csv.ReadOptions());
var frame = new DataFrame(columns, FIndex.Of(names));
Console.WriteLine(Formatter.FrameRepr(frame));
Console.WriteLine(frame.GetColumn((object?)"value"));
if (frame.NRows != 2) throw new Exception("Row count mismatch");
```
