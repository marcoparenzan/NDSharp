# Reproducible random sampling

Use a seeded Generator for repeatable results. Reusing a generator advances its stream; recreating it with the same seed restarts it.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.8.1-net11-preview.1"
#r "nuget: NDSharp.Frame, 2.5.1-net11-preview.1"
#r "nuget: NDSharp.Plot, 2.6.3-net11-preview.2"
#r "nuget: NDSharp.Image, 2.5.1-net11-preview.1"
#r "nuget: NDSharp.Wavelets, 2.5.1-net11-preview.1"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var seed = new System.Numerics.BigInteger[] { 42 };
var rng = NDSharp.Random.Generator.DefaultRng(seed);
var a = rng.Random(new[] { 2, 3 });
Console.WriteLine(a);
Console.WriteLine(rng.StandardNormal(new[] { 5 }));
var repeat = NDSharp.Random.Generator.DefaultRng(seed).Random(new[] { 2, 3 });
if (!(bool)np.All(np.Equal(a, repeat)).GetAt(0)) throw new Exception("Seed mismatch");
```
