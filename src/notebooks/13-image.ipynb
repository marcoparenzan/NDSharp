# OpenCV array conversions

Requires the Windows OpenCV native runtime supplied by NDSharp.Image, or a compatible native deployment on another OS. Image layout is height × width × channels; OpenCV uses BGR order.

Run cells top to bottom with the .NET Interactive C# kernel. See ../../USAGE.md for package-feed setup.

```
#r "nuget: NDSharp, 2.8.1-net11-preview.1"
#r "nuget: NDSharp.Frame, 2.5.1-net11-preview.1"
#r "nuget: NDSharp.Plot, 2.6.3-net11-preview.2"
#r "nuget: NDSharp.Image, 2.5.1-net11-preview.1"
#r "nuget: NDSharp.Wavelets, 2.5.1-net11-preview.1"
using System;
using System.Linq;
using System.IO;
using NDSharp;
using NDSharp.Frame;
using NDSharp.Plot;
using NDSharp.Image;
using NDSharp.Wavelets;
using SkiaSharp;
using FIndex = NDSharp.Frame.Index;
```

```
var bgr = np.Full(new[] { 2, 2, 3 }, 100L, DType.UInt8);
var gray = Cv.CvtColor(bgr, 6); // OpenCV COLOR_BGR2GRAY
Console.WriteLine(gray);
using var mat = Mats.ToMat(gray);
var restored = Mats.FromMat(mat);
if (!(bool)np.All(np.Equal(gray, restored)).GetAt(0)) throw new Exception("Mat roundtrip mismatch");
```
